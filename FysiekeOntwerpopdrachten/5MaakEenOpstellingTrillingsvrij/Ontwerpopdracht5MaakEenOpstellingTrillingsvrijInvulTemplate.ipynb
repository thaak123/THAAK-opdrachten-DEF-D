## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
f0 = 2.5 #Hz
w0 = 2*np.pi*f0

print("De benodigde kantelfrequentie is %.1f Hz" %f0)
```

```python
# Voeg hier je plot in.
f_arr = np.linspace(0,10, 100)
w_arr = 2*np.pi*f_arr
A_quo = 1 / (1 - (w_arr/w0)**2)

plt.figure()
plt.loglog(f_arr, A_quo)
plt.xlabel(r"$f$ [Hz]")
plt.ylabel(r"$\frac{A_{out}}{A_{in}}$ [-]")
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g = 9.81 #m/s^2

delta_x = g / (w0**2)
print("Delta-x is %.3f m" %delta_x)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

import numpy as np
import matplotlib.pyplot as plt

massa = np.linspace(0,5,100)

delta_x_2 = 2*delta_x
delta_x_5 = 5*delta_x

def C_function(m, delta_x):
    return m*g / delta_x

plt.figure()
plt.plot(massa, C_function(massa, delta_x), label="delta_x = a")
plt.plot(massa, C_function(massa, delta_x_2), label="delta_x = 2a")
plt.plot(massa, C_function(massa, delta_x_5), label="delta_x = 5a")
plt.savefig("Opdr3_plot")
plt.xlabel(r"$m$ [kg]")
plt.ylabel(r"$C$ [N/m]")
plt.legend()
plt.show()
```
