![Alt](afleiding.jpg "Schets")

```python
# Bereken hier je benodigde kantelfrequentie
f0 = 2.5 #Hz
w0 = 2*np.pi*f0

print("De benodigde kantelfrequentie is %.1f Hz" %f0)
```

```python
# Voeg hier je plot in.
f_arr = np.linspace(0,10, 100)
w_arr = 2*np.pi*f_arr
A_quo = np.abs(1 / (1 - (w_arr/w0)**2))

plt.figure()
plt.loglog(f_arr, A_quo)
plt.xlabel(r"$f$ [Hz]")
plt.ylabel(r"$\frac{A_{out}}{A_{in}}$ [-]")
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g = 9.81 #m/s^2

delta_x = g / (w0**2)
print("Delta-x is %.3f m" %delta_x)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

import numpy as np
import matplotlib.pyplot as plt

massa = np.linspace(0,5,100)

delta_x_2 = 2*delta_x
delta_x_5 = 5*delta_x

def C_function(m, delta_x):
    return m*g / delta_x

plt.figure()
plt.plot(massa, C_function(massa, delta_x), label="delta_x = a")
plt.plot(massa, C_function(massa, delta_x_2), label="delta_x = 2a")
plt.plot(massa, C_function(massa, delta_x_5), label="delta_x = 5a")
plt.savefig("Opdr3_plot")
plt.xlabel(r"$m$ [kg]")
plt.ylabel(r"$C$ [N/m]")
plt.legend()
plt.show()
```

```python
# Voeg hier je plots in.

# Herhaling van eerdere opdracht
f_arr = np.linspace(0,10, 100)
w_arr = 2*np.pi*f_arr

delta_x_eis = 0.08
w0_eis = np.sqrt(g/delta_x_eis)
f0_eis = w0_eis/(2*np.pi)
A_quo = np.abs(1 / (1 - (w_arr/w0)**2))

delta_x_ontwerp = 0.13
w0_ontwerp = np.sqrt(g/delta_x_ontwerp)
f0_ontwerp = w0_ontwerp/(2*np.pi)
A_quo_ontwerp = np.abs(1 / (1 - (w_arr/w0_ontwerp)**2))

# Metingen
f_arr_metingen = [1.1,2.9,5.1]

# Bij 1.3 Hz
A_out_1 = 1.0
A_in_1 = 0.22
A_1 = np.abs(A_out_1/A_in_1)

#2.9
A_out_3 = 0.92
A_in_3 = 1.33
A_3 = np.abs(A_out_3/A_in_3)


#5.1
A_out_5 = 1.06
A_in_5 = 4.16
A_5 = np.abs(A_out_5/A_in_5)

A_metingen = [A_1,A_3,A_5]
u_f = 0.4
u_A = 0.2

plt.figure()
plt.plot(f_arr, A_quo, label="Grens ontwerp-eis")
plt.plot(f_arr, A_quo_ontwerp, label="Ontwerp")
plt.errorbar(f_arr_metingen, A_metingen, xerr=u_f, yerr=u_A, fmt = 'o',label="Metingen")
plt.xscale('log')
plt.yscale('log')
plt.xlabel(r"$f$ [Hz]")
plt.ylabel(r"$\frac{A_{out}}{A_{in}}$ [-]")
plt.legend()
plt.savefig("meetgrafiek.png")
plt.show()
```
